```python
scaler = MinMaxScaler()
```

```python
def stock_price_prediction(file):
  df=pd.read_csv(file)

  # Determine the correct date column name
  # date_column = None
  # if 'date' in df.columns:
  #   date_column = 'date'
  # elif 'Date' in df.columns:
  #   date_column = 'Date'
  # else:
  #   raise ValueError("CSV must contain a 'date' or 'Date' column.")

  df['date'] = pd.to_datetime(df['date'], format='%d-%m-%Y') # Apply format based on warning
  df = df.sort_values(by='date')

  last_20 = df.tail(20)
  previous_prices = last_20[['open', 'high', 'low', 'close', 'volume', 'rsi_7', 'macd', 'bollinger']].values
  previous_prices_scaled = scaler.fit_transform(previous_prices)
  previous_prices_input = np.reshape(previous_prices_scaled, (1, 20, 8))
  prediction = model.predict(previous_prices_input, verbose=0)

  # Create a dummy array of shape (1, 8) to match the scaler's fitted features
  # Assuming the model predicts the 'close' price, which is at index 3 in the feature list.
  dummy_inverse_input = np.zeros((1, 8))
  dummy_inverse_input[0, 3] = prediction[0, 0] # Place the single predicted value in the 'close' price position

  # Inverse transform the dummy array
  unscaled_dummy_output = scaler.inverse_transform(dummy_inverse_input)

  # Extract the unscaled 'close' price prediction
  prediction = unscaled_dummy_output[0, 3]

  return float(prediction)
```

```python
make_profit = gr.Interface(
    fn = stock_price_prediction,
    inputs=gr.File(label='upload csv with latest 20 days'),
    outputs=gr.Number(label='predicted price'),
    title="Apple stock prediction app",
    description="At your own risk"
)
```
